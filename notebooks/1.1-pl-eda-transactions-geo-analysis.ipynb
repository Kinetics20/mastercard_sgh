# geolocation Analysis of Transactions

Goal: Analyse special patterns of fraud transactions, detect randomness, and engineer useful location features

```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
PROCESSED_DATA = Path("../data/processed")
FIGURES_PATH = Path('../reports/figures')
```

```python
df =pd.read_parquet(PROCESSED_DATA / 'transactions.parquet')
```

```python
df.head()
```

```python
df_location = pd.json_normalize(df['location'])
```

```python
df_location.head()
```

```python
df = pd.concat([df.drop(columns=['location']),df_location], axis=1)
```

```python
print('latitude range:', df['lat'].min(), df['lat'].max())
print('latitude range:', df['long'].min(), df['long'].max())
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))


sns.histplot(df['lat'], bins=50, ax=ax[0], color='blue')
ax[0].set_title('Latitude Distribution')


sns.histplot(df['long'], bins=50, ax=ax[1], color='red')
ax[1].set_title('Longitude Distribution')


fig.suptitle("Raw Geolocation Distribution")
fig.tight_layout()
plt.show()
```

```python
label_map = {0: 'Legitimate', 1: 'Fraudulent'}
df_sampled = df.sample(10000, random_state=42).copy()
df_sampled['fraud_label'] = df_sampled['is_fraud'].map(label_map)
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

sns.scatterplot(
    data = df_sampled,
    ax=ax,
    x= 'long',
    y='lat',
    hue='fraud_label',
    alpha=0.3,
    palette={'Legitimate': 'blue', 'Fraudulent': 'red'},
)

ax.set_title('Transaction Locations (Sampled)')
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")

ax.set_aspect('equal')

ax.legend(title='Transaction Type', loc='center left', bbox_to_anchor=(1, 0.5))

fig.tight_layout()

plt.savefig(FIGURES_PATH / 'transactions_locations_sampled.png')
plt.show()
```

```python
fig = px.scatter_map(
    df_sampled.sample(5000, random_state=42),
    lat='lat',
    lon='long',
    color='fraud_label',
    hover_data=['is_fraud'],
    zoom=3,
    height=650,
    color_discrete_map={'Legitimate': 'rgba(0, 0, 255, 0.3)', 'Fraudulent': 'red'},

)
# from HTML save to png file, the best way to save pict from the map
# fig.write_html(FIGURES_PATH / 'transactions_locations_sampled_map.html')
fig.show()
```
