```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

from mastercard import dist_visualisation, dist_categorical_visualisation
from mastercard.config import INTERIM_DATA_DIR
```

```python
PROCESSED_DATA = Path("../data/processed")
INTERIM_DATA = Path("../data/interim")
FIGURES_PATH = Path('../reports/figures')
```

```python
transactions = pd.read_parquet(PROCESSED_DATA / 'transactions.parquet')
users = pd.read_parquet(PROCESSED_DATA / 'users.parquet')
merchants = pd.read_parquet(PROCESSED_DATA / 'merchants.parquet')
```

```python
users['signup_date'] = pd.to_datetime(users['signup_date'])
```

```python
transactions.head()
```

```python
users.head()
```

```python
merchants.head()
```

```python
df = transactions.merge(users, on='user_id', how='left').merge(merchants, on='merchant_id', how='left', suffixes=('_user', '_merchant'))
```

### Users vs is_fraud

```python
frauds_per_user = df.groupby('user_id')['is_fraud'].sum()
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

frauds_per_user.value_counts().sort_index().plot(ax=ax, kind='bar')

ax.set_xlabel("Count fraud per user")
ax.set_ylabel("Count users")

ax.set_title('Distribution of frauds by user')


fig.tight_layout()
plt.show()
```

```python
transactions_per_user = df.groupby('user_id')['is_fraud'].count()


fig, ax = plt.subplots(figsize=(12, 5))

transactions_per_user.value_counts().sort_index().plot(ax=ax, kind='bar')

ax.set_xlabel("Count transactions per user")
ax.set_ylabel("Count users")

ax.set_title('Distribution of transactions by user')


fig.tight_layout()
plt.show()
```

```python
user_stats = (
    df.groupby('user_id')
    .agg(transactions=('is_fraud', 'count'), frauds=('is_fraud', 'sum'))
)
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

ax.scatter(user_stats['transactions'], user_stats['frauds'], alpha=0.3)
ax.set_xlabel("Transactions Counts per user")
ax.set_xlabel("Frauds Counts per user")

ax.set_title("Relationship between transactions and frauds")
fig.tight_layout()
plt.show()


```

```python
corr = user_stats.corr()
fig, ax = plt.subplots(figsize=(6, 4))

sns.heatmap(
    corr,
    ax=ax,
    annot=True,
    fmt='.2f',
    square=True,
    linewidths=.5,
    cbar_kws={'shrink': 0.8},
    cmap='coolwarm'
)

ax.set_title('Correlation Heatmap', fontsize=14)
plt.tight_layout()
plt.show()
```

### Merchants vs is_fraud

```python
merchants_stats = (
    df.groupby('merchant_id')
    .agg(transactions=('is_fraud', 'count'), frauds=('is_fraud', 'sum'))
)
```

### Category vs is_fraud

```python
df['category'].value_counts()
```

```python
category_stats = (
    df.groupby('category')
    .agg(transactions=('is_fraud', 'count'), frauds=('is_fraud', 'sum'))
    .sort_values('frauds', ascending=False)
)



fig, ax = plt.subplots(figsize=(12, 5))

ax.bar(category_stats.index, category_stats['frauds'], label='Frauds')
ax.bar(category_stats.index, category_stats['transactions'] - category_stats['frauds'], bottom=category_stats['frauds'], label='Transactions')


ax.set_title("Distribution of frauds by category")
ax.set_xlabel("Category")
ax.set_ylabel("Count frauds")
# ax.set_ylim(0, 1000)

fig.tight_layout()

# plt.savefig()
plt.show()
```

```python
import matplotlib.pyplot as plt


category_stats = (
    df.groupby('category')
    .agg(transactions=('is_fraud', 'count'), frauds=('is_fraud', 'sum'))
    .sort_values('frauds', ascending=False)
    .reset_index()
)


category_stats['non_frauds'] = category_stats['transactions'] - category_stats['frauds']


fig, ax = plt.subplots(figsize=(12, 6))


ax.bar(category_stats['category'], category_stats['non_frauds'], label='Non-Fraud', color='skyblue')


ax.bar(category_stats['category'], category_stats['frauds'], bottom=category_stats['non_frauds'], label='Fraud', color='crimson')


ax.set_title("Fraud vs Non-Fraud Transactions by Category")
ax.set_xlabel("Category")
ax.set_ylabel("Number of Transactions")
plt.xticks(rotation=45)
ax.legend()

fig.tight_layout()
plt.show()
```

# Fraud rate international vs local

```python
fraud_rate_by_international = df.groupby('is_international')['is_fraud'].mean().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(12, 5))
fraud_rate_by_international.plot(kind='bar', ax=ax)
ax.set_title('Fraud rate by international')
ax.set_xlabel("International")
ax.set_ylabel('Fraud rate')

ax.set_xticklabels(['International', 'Not International'])
ax.set_xticks(ticks=(0, 1), labels=['International', 'Local'],rotation=0, ha='center')

fig.tight_layout()
plt.show()

```
